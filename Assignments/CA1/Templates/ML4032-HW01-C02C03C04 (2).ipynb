## **Part 1: Provided Dataset Generation**

The following code snippets generate six synthetic datasets using scikit-learn. Each dataset is designed to reveal unique challenges to classifiers.

1. **Blobs (Well-Separated Clusters)**
   
   ```python
   from sklearn.datasets import make_blobs
   X_blobs, y_blobs = make_blobs(n_samples=300, centers=4, cluster_std=1.2, random_state=42)
   ```
   
   *Purpose:* Evaluates classifier performance on isotropic Gaussian clusters with clear separations.

2. **Classification (Informative vs. Redundant Features)**
   
   ```python
   from sklearn.datasets import make_classification
   X_class, y_class = make_classification(n_samples=500, n_features=10,
                                          n_informative=5, n_redundant=2,
                                          n_clusters_per_class=2, random_state=12)
   ```
   
   *Purpose:* Tests robustness when redundant or noisy features are present.

3. **Gaussian Quantiles (Concentric Class Boundaries)**
   
   ```python
   from sklearn.datasets import make_gaussian_quantiles
   X_gaussian, y_gaussian = make_gaussian_quantiles(n_samples=500, n_features=2, n_classes=3, random_state=0)
   ```
   
   *Purpose:* Investigates performance on data with curved, concentrically separated classes.

4. **Circles (Nonlinear, Concentric Patterns)**
   
   ```python
   from sklearn.datasets import make_circles
   X_circles, y_circles = make_circles(n_samples=400, noise=0.1, factor=0.5, random_state=7)
   ```
   
   *Purpose:* Challenges classifiers with nonlinearly separable boundaries that can trip up linear models.

5. **Moons (Interleaving Half-Circles)**
   
   ```python
   from sklearn.datasets import make_moons
   X_moons, y_moons = make_moons(n_samples=400, noise=0.2, random_state=7)
   ```
   
   *Purpose:* Assesses classifier performance on interleaving curved boundaries with added noise.

6. **Hastie 10-2 (Higher-Dimensional Ambiguity)**
   
   ```python
   from sklearn.datasets import make_hastie_10_2
   X_hastie, y_hastie = make_hastie_10_2(n_samples=500, random_state=21)
   ```
   
   *Purpose:* Explores classifier behavior in a higher-dimensional setting where classes overlap.
